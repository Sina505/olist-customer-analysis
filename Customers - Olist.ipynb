<h2>Fixing Nulls:</h2>

<h4>Orders Dataset</h4>

```python
# Examine the relationship between the order_status feature and other three important features
# It is only for calculating delivery time and delay

date_cols= ["order_approved_at", "order_delivered_carrier_date", "order_delivered_customer_date"]
Orders.groupby("order_status")[date_cols].apply(lambda x: x.isnull().sum())

# Null values are expected for undeliverd order, but all three features must be present for delivered orders
```

```python
# finding anomalies in delivered column

delivered= Orders[Orders["order_status"]== "delivered"]
anomalies= delivered[delivered[date_cols].isnull().any(axis=1)]

print(len(anomalies))
anomalies.head(10)
```

```python
# cleaning anomalies

delivered_clean_orders= delivered.dropna(subset=date_cols).copy()
```

```python
# checking the dataset

print(delivered_clean_orders.isnull().sum())

print("\n", 
      Orders.shape[0], "\n",
      delivered.shape[0], "\n",
      delivered_clean_orders.shape[0])
```

<h2>Analysis:</h2>

<h4>Repeat Purchase</h4>

```python
orders_customers= Orders.merge(Customers, on="customer_id", how="left")
```

```python
purches_freq= orders_customers.groupby("customer_unique_id")["order_id"].count()
freq_counts= purches_freq.value_counts().sort_index()

freq_top5= freq_counts.loc[freq_counts.index <=5]
```

```python
# colors

colors_repeat_p= []

for count in freq_top5.index:
    
    if count == 1:
        colors_repeat_p.append("red")
        
    elif 2 <= count <= 3:
        colors_repeat_p.append("orange")
        
    else:
        colors_repeat_p.append("green")
```

```python
plt.figure(figsize=(8,5))

plt.bar(freq_top5.index, freq_top5.values, color=colors_repeat_p)
plt.title("Number of customers by purchase frequency (up to 5)")
plt.xlabel("Number of purchase")
plt.ylabel("Number of customers")
plt.xticks(freq_top5.index)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(order_value[order_value < order_value.quantile(0.95)], bins= 50, kde=True)
plt.title("Order value distribution")
plt.xlabel("Order value")
plt.axvline(order_value.mean(), color="r", linestyle="--", label="Average")

plt.legend()
plt.show()
```

<h4>Order delivery time distribution</h4>

```python
# Calculates the time difference between purchase and actual delivery, in days

delivered_clean_orders["order_delivered_customer_date"]= pd.to_datetime(delivered_clean_orders["order_delivered_customer_date"])
delivered_clean_orders["order_purchase_timestamp"]= pd.to_datetime(delivered_clean_orders["order_purchase_timestamp"])

delivered_clean_orders["delivery_time_days"]= (delivered_clean_orders["order_delivered_customer_date"] -
                                               delivered_clean_orders["order_purchase_timestamp"]).dt.days
```

```python
plt.figure(figsize=(8,5))

sns.histplot(delivered_clean_orders["delivery_time_days"], bins=45, kde=True)

plt.axvline(delivered_clean_orders["delivery_time_days"].mean(), color="r", linestyle="--", label="Average")
plt.title("Order delivery time distribution")
plt.xlabel("Delivery time (Days)")
plt.ylabel("Number of orders")

plt.legend()
plt.show()
```

<h4>Average payment value by number of installments</h4>

```python
payments_filtered= Payments[Payments["payment_installments"] > 0]

avg_value_installments= payments_filtered.groupby("payment_installments")["payment_value"].mean()
```

```python
# colors

colors_installments= []

for installments, value in avg_value_installments.items():
    
    if installments >= 10 and value > 400:
        colors_installments.append("green")
        
    elif 1 <= installments <= 7 and value < 300:
        colors_installments.append("green")
        
    elif 8 <= installments <= 15 and value >= 200:
        colors_installments.append("orange")
        
    else:
        colors_installments.append("red")
```

```python
plt.figure(figsize=(10,5))

plt.bar(avg_value_installments.index, avg_value_installments.values, color= colors_installments)
plt.title("Average payment value by number of installments")
plt.xlabel("Number of installments")
plt.ylabel("Avarage payment value")

plt.show()
```

<h4>Heatmap</h4>

```python
order_value = Order_items.groupby('order_id')['price'].sum().reset_index()
order_review = Reviews.groupby('order_id')['review_score'].mean().reset_index()
order_payment = Payments.groupby('order_id')['payment_value'].sum().reset_index()
order_installments = Payments.groupby('order_id')['payment_installments'].max().reset_index()

corr_df = order_value.merge(order_review, on='order_id')
corr_df = corr_df.merge(order_payment, on='order_id')
corr_df = corr_df.merge(order_installments, on='order_id')
corr_df = corr_df.merge(delivered_clean_orders[['order_id', 'delivery_time_days']], on='order_id')
```

```python
corr_matrix = corr_df[['price', 'review_score', 'payment_value', 'payment_installments', 'delivery_time_days']].corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr_matrix, annot=True, cmap='summer')
plt.title('Correlation heatmap')
plt.show()
```
